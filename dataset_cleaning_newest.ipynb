```python
df['main_category'] = df['main_category'].astype(str)
df['sub_category'] = df['sub_category'].astype(str)
```

```python
non_numeric_ratings = df[pd.to_numeric(df['ratings'], errors='coerce').isnull()]
non_numeric_ratings
```

```python
# Define a function to replace specific strings with NaN
def replace_with_nan(x):
    if isinstance(x, str) and (
        x.startswith("Get") or x.startswith("Free") or x.startswith("FREE") or x.startswith("₹")
    ):
        return np.nan
    else:
        return x

# Apply the function to the 'ratings' column
df['ratings'] = df['ratings'].apply(replace_with_nan)

# Verify the changes
df['ratings'].head()
```

```python
df['ratings'] = df['ratings'].astype(float)
df.info()
```

```python
df.head(10)
```

```python
# Find rows where 'no_of_ratings' contains non-numeric values
non_numeric_no_of_ratings = df[pd.to_numeric(df['no_of_ratings'], errors='coerce').isnull()]

# Display rows with non-numeric values in 'no_of_ratings'
non_numeric_no_of_ratings
```

```python
# Define a function to replace specific strings with random integers
def replace_with_nan_no_of_ratings(x):
    if isinstance(x, str) and (
        x.startswith("Usually dispatched") or x.startswith("Only") or x.startswith("Free") or x.startswith("This item") or x.startswith("FREE Delivery")
    ):
        return np.nan
    else:
        return x

# Apply the function to the 'no_of_ratings' column
df['no_of_ratings'] = df['no_of_ratings'].apply(replace_with_nan_no_of_ratings)
```

```python
df['no_of_ratings'] = df['no_of_ratings'].str.replace(',', '')
df.head(10)
```

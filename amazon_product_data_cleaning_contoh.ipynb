### cek jika ada missing value ada no_of_ratings

```python
nan_no_of_ratings = df[df['no_of_ratings'].isna()]
print(nan_no_of_ratings)
```

```python
df['no_of_ratings'].tail()
```

```python
df['no_of_ratings'] = df['no_of_ratings'].str.replace(',', '')
```

```python
# Define a function to replace specific strings with random integers
def replace_with_random_int(x):
    if isinstance(x, str) and (
        x.startswith("Usually dispatched") or x.startswith("Only") or x.startswith("Free") or x.startswith("This item") or x.startswith("FREE Delivery")
    ):
        return np.random.randint(1, 2501)
    else:
        return x

# Apply the function to the 'no_of_ratings' column
df['no_of_ratings'] = df['no_of_ratings'].apply(replace_with_random_int)
```

```python
df['no_of_ratings'] = pd.to_numeric(df['no_of_ratings'], errors='coerce')
```

```python
df['no_of_ratings'] = df['no_of_ratings'].apply(lambda x: np.random.randint(1, 2501) if pd.isna(x) else x)
```

```python
df['no_of_ratings'] = df['no_of_ratings'].astype(int)
```

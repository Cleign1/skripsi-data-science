```python
import pandas as pd
import glob
```

```python
# Tentukan folder berisi semua file daily_sales
file_pattern = 'daily_sales/daily_sales_202504*.csv'

# Ambil semua nama file sesuai pattern
file_list = glob.glob(file_pattern)
```

```python
df = pd.concat([pd.read_csv(f, parse_dates=['date']) for f in file_list], ignore_index=True)
```

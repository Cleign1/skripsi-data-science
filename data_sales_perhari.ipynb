```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
```

```python
produk_df = pd.read_csv('dataset/amazon_product_cleaned.csv')
produk_df.head(10)
```

```python
start_date = datetime.strptime("2025-04-01", "%Y-%m-%d")
end_date = datetime.strptime("2025-04-15", "%Y-%m-%d")
```

```python
delta_hari = (end_date - start_date).days + 1
```

```python
for i in range(delta_hari):
    current_date = start_date + timedelta(days=i)
    daily_sales = produk_df[['index', 'name']].copy()
    daily_sales['date'] = current_date.strftime('%Y-%m-%d')
    
    # generate random sales data 
    daily_sales['quantity_sold'] = np.random.randint(1, 50, size=len(daily_sales))
    
    filename = f'daily_sales/daily_sales_{current_date.strftime("%Y%m%d")}.csv'
    daily_sales.to_csv(filename, index=False)
    
    print(f"File {filename} created with {len(daily_sales)} records.")
```
